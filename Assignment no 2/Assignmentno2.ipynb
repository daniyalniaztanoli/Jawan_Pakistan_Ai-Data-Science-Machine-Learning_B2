```python
# Question no 3

course_a = {"Daniyal" , "Noman" , "A.rehman" , "Bilal"}
course_b = {"Danish" , "Noman" , "Rehman" , "Bilal"}

difference = course_a.difference(course_b)
symmetric = course_a.symmetric_difference(course_b)

print("Students in Course A but not in Course B" , difference)
print("Students are only in one of two course " , symmetric)
```
