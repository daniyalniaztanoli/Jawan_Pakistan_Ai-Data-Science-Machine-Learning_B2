```python
# Question no 1
products = {
    101: {"name": "Laptop", "category": "Electronics", "price": 1200, "stock": 50},
    102: {"name": "Shirt", "category": "Apparel", "price": 25, "stock": 200},
    103: {"name": "Coffee Maker", "category": "Home Appliances", "price": 80, "stock": 30}
}

products[102]["stock"] = 250

print("Updated Products:")
print(products)

products[104] = {
    "name": "Smartphone",
    "category": "Electronics",
    "price": 800,
    "stock": 100
}

print("\nAfter Adding New Product:")
print(products)
```

```python
# Question no 2
age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age")
elif age <= 12:
    print("Child")
elif age <= 19:
    print("Teenager")
elif age <= 64:
    print("Adult")
else:
    print("Senior")
```

```python
# Question no 3
title = input("Enter book title: ")
author = input("Enter author's name: ")
year = int(input("Enter year of publication: "))
price = float(input("Enter book price: "))

book = (title, author, year, price)

print("\nBook Details:")
print("Title:", book[0])
print("Author:", book[1])
print("Year of Publication:", book[2])
print(f"Price: ${book[3]:.1f}")
```

```python
# Question no 8
inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 8},
    "Harry Potter": {"price": 12.50, "quantity": 2}
}


inventory.update({
    "Atomic Habits": {"price": 15.99, "quantity": 4}
})

print("All Book Titles:")
print(list(inventory.keys()))

for book in list(inventory.keys()):
    if inventory[book]["quantity"] == 0:
        inventory.pop(book)

print("\nUpdated Inventory:")
print(inventory)
```

```python
# Question no 10
letter = input("Enter a single letter: ")

if len(letter) == 1 and letter.isalpha():
    if letter.lower() in "aeiou":
        print(letter, "is a Vowel")
    else:
        print(letter, "is a Consonant")
else:
    print("Please enter valid alphabet")
```
